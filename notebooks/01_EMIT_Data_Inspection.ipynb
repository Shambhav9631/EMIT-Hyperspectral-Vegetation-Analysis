# 01 — EMIT Data Inspection

Inspect an EMIT L2A reflectance NetCDF file before analysis.

**Input:** EMIT `.nc` file  
**Output:** dataset summary, dimensions, variables, coordinates, metadata, and wavelength checks.

```python
from pathlib import Path
import xarray as xr
import numpy as np

DATA_DIR = Path("../data")
EMIT_NC = DATA_DIR / "EMIT_L2A_RFL_example.nc"

if not EMIT_NC.exists():
    raise FileNotFoundError(
        f"Place your EMIT NetCDF file at: {EMIT_NC.resolve()}"
    )

ds = xr.open_dataset(EMIT_NC)
print(ds)
```

```python
print("===== DIMENSIONS =====")
for dim, size in ds.sizes.items():
    print(f"{dim}: {size}")

print("\n===== DATA VARIABLES =====")
for var in ds.data_vars:
    print(var)

print("\n===== COORDINATES =====")
for coord in ds.coords:
    print(coord)
```

```python
print("===== REFLECTANCE =====")
if "reflectance" in ds:
    print(ds["reflectance"])
    print("Shape:", ds["reflectance"].shape)
    print("Dtype:", ds["reflectance"].dtype)
else:
    print("reflectance variable not found")

print("\n===== WAVELENGTH =====")
if "wavelength" in ds:
    wl = np.asarray(ds["wavelength"].values)
    print("Count:", len(wl))
    print("First:", wl[:10])
    print("Last:", wl[-10:])
else:
    print("wavelength variable not found")
```

```python
print("\n===== GLOBAL ATTRIBUTES =====")
for key, value in ds.attrs.items():
    print(f"{key}: {value}")
```

```python
# EMIT files may contain a location group.
try:
    location = xr.open_dataset(EMIT_NC, group="location")
    print("===== LOCATION GROUP =====")
    print(location)
    print("\nVariables:", list(location.data_vars))
except Exception as exc:
    print("Location group could not be opened:", exc)
```
